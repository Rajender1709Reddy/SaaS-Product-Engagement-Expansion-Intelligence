```python
customers = customers.drop_duplicates()
product = product.drop_duplicates()
```

```python
customers.columns = customers.columns.str.strip()
product.columns = product.columns.str.strip()
```

```python
for col in customers.select_dtypes(include="object"):
    customers[col] = customers[col].str.strip()

for col in product.select_dtypes(include="object"):
    product[col] = product[col].str.strip()
```

```python
## Convert Customer Dte Columns
customer_dates = [
    "signup_date",
    "first_login_date",
    "last_login_date"
]

for col in customer_dates:
    customers[col] = pd.to_datetime(
        customers[col],
        format="%Y-%m-%d",
        errors="coerce"
    )
```

```python
## convert Product Date columns 
product_dates = [
    "subscription_start_date",
    "subscription_end_date",
    "first_feature_used",
    "usage_date",
    "ticket_created_date",
    "ticket_closed_date"
]

for col in product_dates:
    product[col] = pd.to_datetime(
        product[col],
        format="%Y-%m-%d",
        errors="coerce"
    )
```

```python
## Fill Missing VAlues 
customers.fillna(customers.median(numeric_only=True), inplace=True)
customers.fillna("Unknown", inplace=True)
```

```python
product.fillna(product.median(numeric_only=True), inplace=True)
product.fillna("Unknown", inplace=True)
```

```python
print("Customer Dataset Shape :", customers.shape)
print("Product Dataset Shape :", product.shape)
```

```python
customers.to_csv(
    r"C:\Users\SVS\Desktop\Product_Adoption_customer_sucess\data\processed\customer_profile_processed.csv",
    index=False
)

product.to_csv(
    r"C:\Users\SVS\Desktop\Product_Adoption_customer_sucess\data\processed\product_operations_processed.csv",
    index=False
)
```
